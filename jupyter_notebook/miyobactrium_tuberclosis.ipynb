Getting the protein FASTA sequence

```python
search = Entrez.esearch(
    db="protein",
    term="Mycobacterium tuberculosis[Organism]",
    retmax=5
)

search_results = Entrez.read(search)

protein_ids = search_results["IdList"]

# Step 2: Get summaries
summary = Entrez.esummary(
    db="protein",
    id=",".join(protein_ids)
)

summary_results = Entrez.read(summary)

# Step 3: Print ID and title
for record in summary_results:
    print("ID:", record["Id"])
    print("Title:", record["Title"])
    print("-" * 50)
```
