## Searching for microbes (bacteria)+ gene compositon


```python
handle = Entrez.esearch(db="taxonomy", term="bacteria[Organism]", retmax=1000)
tax_records = Entrez.read(handle)
handle.close()

print(tax_records["IdList"])
```

## Getting details of each microbes


```python
for tax_id in tax_records["IdList"]:
    handle = Entrez.efetch(db="taxonomy", id=tax_id, retmode="xml")
    data = Entrez.read(handle)
    handle.close()
    
    print("Name:", data[0]["ScientificName"])
    print("Lineage:", data[0]["Lineage"])
    print("-" * 40)
```

## Getting genes inside a microbe gene composition


```python
organism = "Escherichia coli"

handle = Entrez.esearch(
    db="gene",
    term=f"{organism}[Organism]",
    retmax=5
)
gene_records = Entrez.read(handle)
handle.close()

print(gene_records["IdList"])
```

## Fetching gene details


```python
for gene_id in gene_records["IdList"]:
    handle = Entrez.efetch(db="gene", id=gene_id, retmode="xml")
    gene_data = Entrez.read(handle)
    handle.close()
    
    gene_info = gene_data[0]
    
    print("Gene:", gene_info["Entrezgene_gene"]["Gene-ref"]["Gene-ref_locus"])
```
